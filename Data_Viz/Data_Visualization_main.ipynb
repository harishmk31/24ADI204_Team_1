```python
df.columns = [
    "List_Sr_No",
    "Set_No",
    "Set_Code",
    "First_Name",
    "Surname",
    "Country",
    "State_Association",
    "Date_of_Birth",
    "Age",
    "Role",
    "Batting_Style",
    "Bowling_Style",
    "Test_Caps",
    "ODI_Caps",
    "T20_Caps",
    "IPL_Matches",
    "Previous_IPL_Teams",
    "Current_Team",
    "No_of_Matches",
    "Cap_Status",
    "Reserve_Price"
]

df.head()
```

```python
df = df.drop(index=[0, 1]).reset_index(drop=True)

print("Shape After Removing First Two Rows:", df.shape)
df.head()
```

```python
df["Cap_Status"] = df["Cap_Status"].astype(str)

# If Cap_Status contains numbers, shift them to No_of_Matches
mask = df["Cap_Status"].str.isnumeric()

df.loc[mask, "No_of_Matches"] = df.loc[mask, "Cap_Status"]
df.loc[mask, "Cap_Status"] = np.nan

df.head()
```

```python
df["Cap_Status"] = df["Cap_Status"].str.strip().str.lower()

df["Cap_Status"] = df["Cap_Status"].replace({
    "capped": "Capped",
    "uncapped": "Uncapped"
})
```

```python
caps_columns = ["Test_Caps", "ODI_Caps", "T20_Caps", "IPL_Matches", "No_of_Matches"]

for col in caps_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
indian_mask = df["Country"] == "India"

df.loc[indian_mask, "State_Association"] = df.loc[indian_mask, "State_Association"].fillna(
    df.loc[indian_mask, "State_Association"].mode()[0]
)

df.loc[~indian_mask, "State_Association"] = df.loc[~indian_mask, "State_Association"].fillna("Not Applicable")
```

```python
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
import pandas as pd
import numpy as np
import re

def convert_price(price):
    if pd.isna(price):
        return np.nan

    price = str(price)
    number = re.findall(r"\d+\.?\d*", price)

    if len(number) == 0:
        return np.nan

    number = float(number[0])

    if "Cr" in price:
        return number * 100
    elif "Lakh" in price:
        return number
    else:
        return number


df["Reserve_Price_Numeric"] = df["Reserve_Price"].apply(convert_price)


df["Reserve_Price_Numeric"] = df["Reserve_Price_Numeric"].fillna(
    df["Reserve_Price_Numeric"].median()
)
```

```python
print("Duplicate Rows Before:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate Rows After:", df.duplicated().sum())
```

```python
Q1 = df["Reserve_Price_Numeric"].quantile(0.25)
Q3 = df["Reserve_Price_Numeric"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Reserve_Price_Numeric"] < lower) |
              (df["Reserve_Price_Numeric"] > upper)]

print("Number of Outliers Detected:", len(outliers))
```

```python
print("Final Shape:", df.shape)
print("\nRemaining Missing Values:")
print(df.isnull().sum())

df.to_csv("cricket_cleaned.csv", index=False)

print("Cleaned Dataset Saved Successfully")
```

## Feature Engineering

```python

df["Age_Group"] = pd.cut(df["Age"],
                        bins=[0, 20, 25, 30, 40],
                        labels=["Young", "Prime", "Experienced", "Senior"])

df["Total_Caps"] = df["Test_Caps"] + df["ODI_Caps"] + df["T20_Caps"]

df["Experience_Level"] = pd.cut(df["Total_Caps"],
                               bins=[0, 10, 50, 200],
                               labels=["New", "Intermediate", "Expert"])

df["Player_Type"] = df["Country"].apply(lambda x: "Indian" if x == "India" else "Overseas")
```

```python
import re
import numpy as np

def convert_price(price):
    if pd.isna(price):
        return np.nan

    price = str(price)
    number = re.findall(r"\d+\.?\d*", price)

    if len(number) == 0:
        return np.nan

    number = float(number[0])

    if "Cr" in price:
        return number * 100
    elif "Lakh" in price:
        return number
    else:
        return number

df["Reserve_Price"] = df["Reserve_Price"].apply(convert_price)
```

```python
df["Reserve_Price"].fillna(df["Reserve_Price"].median(), inplace=True)
df["Age"].fillna(df["Age"].median(), inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["Age_Group"] = le.fit_transform(df["Age_Group"])
df["Experience_Level"] = le.fit_transform(df["Experience_Level"])
```

```python
df = pd.get_dummies(df, columns=[
    "Country",
    "Role",
    "Batting_Style",
    "Bowling_Style",
    "Player_Type"
], drop_first=True)
```

```python
import numpy as np

df["Reserve_Price"] = np.log1p(df["Reserve_Price"])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

cols_to_scale = ["Age", "Total_Caps", "IPL_Matches", "No_of_Matches", "Reserve_Price"]

df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

```python
# Feature
df["Total_Caps"] = df["Test_Caps"] + df["ODI_Caps"] + df["T20_Caps"]

# Convert price
df["Reserve_Price"] = df["Reserve_Price"].apply(convert_price)

# Fix skew
df["Reserve_Price"] = np.log1p(df["Reserve_Price"])

# Scale
df[["Reserve_Price"]] = StandardScaler().fit_transform(df[["Reserve_Price"]])
```

```python
# Drop non-useful columns (names, IDs, etc.)
df_model = df.drop([
    "List_Sr_No",
    "First_Name",
    "Surname",
    "Date_of_Birth",
    "Set_Code"
], axis=1)
```

```python
df_numeric = df_model.select_dtypes(include=[np.number])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_numeric)
```

```python
from sklearn.decomposition import PCA

pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

```python
explained_variance = pca.explained_variance_ratio_
print(explained_variance)
```

```python
import numpy as np

cumulative_variance = np.cumsum(explained_variance)
print(cumulative_variance)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(cumulative_variance, marker='o')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Variance Explained")
plt.grid()
plt.show()
```

```python
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_scaled)

print("Selected Components:", pca.n_components_)
```

```python
plt.figure()
plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Projection")
plt.show()
```
